```python
import os
import shutil
import subprocess
from pathlib import Path

def create_lmdb(
    image_directory_original,
    gt_file_original,
    lmdb_output_dir,
    script_path="create_lmdb_dataset.py",
):
    """
    Safe LMDB creation function for Google Colab.

    Expected folder structure:
        dataset_root/
        ├── gt.txt
        └── images/
            ├── 0_xxx.jpg / png ...
            ├── 1_xxx.jpg
            └── ...

    Parameters
    ----------
    image_directory_original : str or Path
        Either:
        (A) the dataset root directory (contains gt.txt + images/), OR
        (B) the "images/" directory.
        The function automatically infers the correct dataset root.

    gt_file_original : str or Path
        Path to gt.txt (usually dataset_root/gt.txt).

    lmdb_output_dir : str or Path
        Output LMDB directory (will be deleted & recreated automatically).

    script_path : str
        Path to create_lmdb_dataset.py (must exist).
    """

    image_directory_original = Path(image_directory_original)
    gt_file_original = Path(gt_file_original)
    lmdb_output_dir = Path(lmdb_output_dir)
    script_path = Path(script_path)

    # --------------------------------------------------
    # (0) Ensure LMDB creation script exists
    # --------------------------------------------------
    if not script_path.is_file():
        raise FileNotFoundError(
            f"ERROR: create_lmdb_dataset.py not found at: {script_path.resolve()}"
        )

    # --------------------------------------------------
    # (1) Infer dataset root folder
    # --------------------------------------------------
    # Case A: user passed dataset_root
    if (image_directory_original / "images").is_dir() and \
       (image_directory_original / "gt.txt").is_file():
        dataset_root = image_directory_original

    # Case B: user passed the images/ folder
    elif image_directory_original.name == "images" and \
         (image_directory_original.parent / "gt.txt").is_file():
        dataset_root = image_directory_original.parent

    else:
        raise ValueError(
            f"ERROR: Could not infer dataset root from: {image_directory_original}\n"
            "Expected folder structure:\n"
            "    dataset_root/\n"
            "        gt.txt\n"
            "        images/\n"
        )

    images_dir = dataset_root / "images"
    gt_inferred = dataset_root / "gt.txt"

    if not images_dir.is_dir():
        raise FileNotFoundError(f"ERROR: images/ folder not found: {images_dir}")

    if not gt_inferred.is_file():
        raise FileNotFoundError(f"ERROR: gt.txt not found: {gt_inferred}")

    # If user passed a different gt path, warn but continue
    if gt_file_original.resolve() != gt_inferred.resolve():
        print(f"WARNING: Provided gt_file = {gt_file_original}")
        print(f"         But dataset root contains gt.txt = {gt_inferred}")
        print("         Will proceed with the provided gt_file_original.\n")

    # --------------------------------------------------
    # (2) Sanity check: count images and gt lines
    # --------------------------------------------------
    valid_exts = {".png", ".jpg", ".jpeg", ".tif", ".tiff", ".bmp"}
    image_files = sorted([p for p in images_dir.iterdir() if p.suffix.lower() in valid_exts])
    num_images = len(image_files)

    if not gt_file_original.is_file():
        raise FileNotFoundError(f"ERROR: Provided gt_file_original does not exist: {gt_file_original}")

    with gt_file_original.open("r", encoding="utf-8", errors="ignore") as f:
        gt_lines = [line.strip() for line in f if line.strip()]
    num_gt = len(gt_lines)

    print("=== Sanity Check ===")
    print(f"Dataset root : {dataset_root}")
    print(f"Images dir   : {images_dir}")
    print(f"# of images  : {num_images}")
    print(f"GT file      : {gt_file_original}")
    print(f"# of GT lines: {num_gt}")

    if num_images != num_gt:
        print("WARNING: #images != #gt lines. Please verify alignment.\n")

    # --------------------------------------------------
    # (3) Prepare output LMDB directory (safe overwrite)
    # --------------------------------------------------
    if lmdb_output_dir.exists():
        print(f"WARNING: Output directory exists → removing: {lmdb_output_dir}")
        shutil.rmtree(lmdb_output_dir)

    lmdb_output_dir.mkdir(parents=True, exist_ok=True)

    # --------------------------------------------------
    # (4) Run create_lmdb_dataset.py
    # --------------------------------------------------
    cmd = [
        "python3",
        str(script_path),
        str(dataset_root),       # IMPORTANT: pass the dataset root, not images/
        str(gt_file_original),
        str(lmdb_output_dir),
    ]

    print("\n=== Running LMDB Creation Command ===")
    print(" ".join(cmd), "\n")

    result = subprocess.run(cmd, stdout=subprocess.PIPE, stderr=subprocess.PIPE, text=True)

    # --------------------------------------------------
    # (5) Print script output
    # --------------------------------------------------
    if result.stdout:
        print("STDOUT:\n" + result.stdout)

    if result.stderr:
        print("STDERR:\n" + result.stderr)

    if result.returncode != 0:
        raise RuntimeError(f"ERROR: create_lmdb_dataset.py failed. Return code = {result.returncode}")

    print("=== LMDB Creation Completed Successfully ===")
    print(f"Output LMDB directory: {lmdb_output_dir.resolve()}")
```

```python
# Scorers
import cv2
import numpy as np
from skimage.morphology import skeletonize

def normalize_score(x, min_val=0, max_val=1):
    """Clamp and normalize to [0,1]."""
    return float(np.clip((x - min_val) / (max_val - min_val + 1e-8), 0, 1))


def calculate_sharp_score(img):
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY) if img.ndim == 3 else img.copy()

    gx = cv2.Sobel(gray, cv2.CV_32F, 1, 0, ksize=3)
    gy = cv2.Sobel(gray, cv2.CV_32F, 0, 1, ksize=3)
    tenengrad_score = float(np.mean(gx**2 + gy**2))

    f = np.fft.fft2(gray)
    fshift = np.fft.fftshift(f)
    mag = np.abs(fshift)
    h, w = mag.shape
    c = 10
    center = mag[h//2 - c:h//2 + c, w//2 - c:w//2 + c]
    high = mag.sum() - center.sum()
    fft_ratio = float(high / (mag.sum() + 1e-8))

    _, bw = cv2.threshold(gray, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU)
    if np.mean(bw) > 127:
        bw = 255 - bw
    skel = skeletonize(bw // 255).astype(np.uint8)
    num_labels, _, stats, _ = cv2.connectedComponentsWithStats(skel, connectivity=8)
    if num_labels > 1:
        lengths = stats[1:, cv2.CC_STAT_AREA]
        continuity = float(np.mean(lengths) / (np.std(lengths) + 1e-6))
    else:
        continuity = 0.0

    # normalize sub-scores
    t_norm = np.tanh(np.log1p(tenengrad_score) / 5.0)
    f_norm = normalize_score(fft_ratio, 0, 1)
    c_norm = np.tanh(continuity / 10.0)

    combined = 0.6 * t_norm + 0.3 * f_norm + 0.1 * c_norm
    return normalize_score(combined, 0, 1)


def calculate_low_light_score(img):
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY) if img.ndim == 3 else img.copy()
    meanY = gray.mean()
    dark_ratio = np.mean(gray < 50)
    alpha, beta = 0.7, 0.3
    score = alpha * (1 - meanY / 255.0) + beta * dark_ratio
    return normalize_score(score, 0, 1)


def calculate_contrast_score(img):
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY) if img.ndim == 3 else img.copy()
    p10 = np.percentile(gray, 10)
    p90 = np.percentile(gray, 90)
    contrast = (p90 - p10) / 255.0
    return normalize_score(contrast, 0, 1)


def calculate_noise_score(img, edge_exclude=0.20, ref_sigma=12.0):
    if img.dtype != np.uint8:
        img = np.clip(img, 0, 255).astype(np.uint8)
    gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY) if img.ndim == 3 else img.copy()

    gx = cv2.Sobel(gray, cv2.CV_32F, 1, 0, ksize=3)
    gy = cv2.Sobel(gray, cv2.CV_32F, 0, 1, ksize=3)
    mag = np.sqrt(gx * gx + gy * gy)

    if edge_exclude > 0:
        thr = np.percentile(mag, 100 * (1 - edge_exclude))
        flat_mask = (mag < thr)
    else:
        flat_mask = np.ones_like(gray, dtype=bool)

    g = gray.astype(np.float32)
    flat_mask &= (g > 5) & (g < 250)
    if flat_mask.mean() < 0.10:
        flat_mask = np.ones_like(gray, dtype=bool)

    hp = g - cv2.GaussianBlur(g, (0, 0), 1.0)
    hp_flat = hp[flat_mask]
    mad = np.median(np.abs(hp_flat - np.median(hp_flat)))
    sigma_mad = 1.4826 * mad

    ker = np.array([[1, -2, 1],
                    [-2, 4, -2],
                    [1, -2, 1]], dtype=np.float32)
    L = cv2.filter2D(g, cv2.CV_32F, ker)[1:-1, 1:-1]
    L_use = np.abs(L[flat_mask[1:-1, 1:-1]])
    sigma_immer = (np.sqrt(np.pi / 2.0) / 6.0) * np.mean(L_use) if L_use.size > 0 else sigma_mad

    sigma_est = float(np.median([sigma_mad, sigma_immer]))

    # normalize noise: low noise=0, high noise=1
    norm_noise = sigma_est / (ref_sigma + sigma_est)
    return normalize_score(norm_noise, 0, 1)
```

```python
# Computing Scorers
import os
import re
import cv2
import pandas as pd
from tqdm import tqdm

def compute_quality_scores(img_dir, out_csv):
    """
    Compute image quality metrics (sharp, noise, contrast, low-light)
    for all images in a folder and save them as a CSV.

    Args:
        img_dir (str): Path to directory with images.
        out_csv (str): Output CSV path.
    """
    # === Ensure output folder exists ===
    os.makedirs(os.path.dirname(out_csv), exist_ok=True)

    # === Helper: natural numeric sort ===
    def natural_key(name):
        return [int(t) if t.isdigit() else t.lower() for t in re.split(r"(\d+)", name)]

    # === Collect and sort image files ===
    images = sorted(
        [f for f in os.listdir(img_dir) if f.lower().endswith((".png", ".jpg", ".jpeg"))],
        key=natural_key
    )
    print(f"📁 Found {len(images)} images in {img_dir}")

    rows = []
    for fname in tqdm(images, desc=f"Analyzing {os.path.basename(img_dir)}"):
        img_path = os.path.join(img_dir, fname)
        img = cv2.imread(img_path)
        if img is None:
            print(f"⚠️ Skipped unreadable image: {fname}")
            continue

        sharp = calculate_sharp_score(img)
        low_light = calculate_low_light_score(img)
        contrast = calculate_contrast_score(img)
        noise = calculate_noise_score(img)

        # assume filename starts with "idx_label.png"
        try:
            idx = int(os.path.splitext(fname)[0].split("_")[0])
        except ValueError:
            print(f"⚠️ Skipped malformed filename: {fname}")
            continue

        rows.append({
            "idx": idx,
            "filename": fname,
            "sharp_score": sharp,
            "low_light_score": low_light,
            "contrast_score": contrast,
            "noise_score": noise,
        })

    # === Save to CSV ===
    df = pd.DataFrame(rows)
    df = df.sort_values("idx").reset_index(drop=True)
    df.to_csv(out_csv, index=False)
    print(f"✅ Saved quality scores for {len(df)} images → {out_csv}")

    return df
```

```python
import pandas as pd
import os

def combine_results_and_scores(detailed_csv, score_csv, out_dir=None, tag=None):
    """
    Combine recognition results and quality scores CSVs by `idx`,
    save both full and NED<1 filtered versions, and return the merged DataFrames.

    Args:
        detailed_csv (str): Path to detailed results CSV (from test.py).
        score_csv (str): Path to image quality scores CSV.
        out_dir (str): Directory to save combined CSVs (default = same as detailed_csv).
        tag (str): Optional dataset name for output filenames.

    Returns:
        tuple[pd.DataFrame, pd.DataFrame]:
            (merged_full, merged_filtered)
    """
    # --- Load both CSVs ---
    df_results = pd.read_csv(detailed_csv)
    df_scores = pd.read_csv(score_csv)

    # --- Ensure correct types for merging ---
    df_results["idx"] = df_results["idx"].astype(int)
    df_scores["idx"] = df_scores["idx"].astype(int)

    # --- Merge ---
    merged = pd.merge(df_results, df_scores, on="idx", how="inner")
    print(f"✅ Combined {len(merged)} records from {os.path.basename(detailed_csv)} and {os.path.basename(score_csv)}")

    # --- Prepare output paths ---
    if out_dir is None:
        out_dir = os.path.dirname(detailed_csv)
    os.makedirs(out_dir, exist_ok=True)
    if tag is None:
        tag = os.path.basename(detailed_csv).replace("_detailed_results.csv", "")

    out_full = os.path.join(out_dir, f"{tag}_combined.csv")
    out_filtered = os.path.join(out_dir, f"{tag}_combined_ned_lt1.csv")

    # --- Save results ---
    merged.to_csv(out_full, index=False)
    merged_filtered = merged[merged["norm_edit_distance"] < 1.0].copy()
    merged_filtered.to_csv(out_filtered, index=False)

    print(f"💾 Saved: {out_full}\n💾 Saved filtered (<1.0 NED): {out_filtered}")

    return merged, merged_filtered
```

## ❇️DOC Denoise Function

```python
import os
from pathlib import Path

import cv2
import numpy as np
from tqdm import tqdm


# ------------------------------
#  Utility: Ensure text = black
# ------------------------------
def _ensure_text_is_black(bin_img):
    """
    確保文字為 0 (黑)，背景為 255 (白)。
    """
    num_black = np.sum(bin_img == 0)
    num_white = np.sum(bin_img == 255)
    if num_black > num_white:
        bin_img = 255 - bin_img
    return bin_img


# -------------------------------------
#  Utility: Compute Otsu threshold on CC sizes
# -------------------------------------
def _compute_otsu_threshold_on_component_sizes(
    sizes,
    norm_max=5000,
):
    sizes = np.asarray(sizes, dtype=np.float64)
    if sizes.size == 0:
        return None

    min_val = sizes.min()
    max_val = sizes.max()
    if max_val <= min_val:
        return None

    norm_indices = ((sizes - min_val) * (norm_max - 1) /
                    (max_val - min_val)).astype(np.int64)

    hist = np.bincount(norm_indices, minlength=norm_max)

    indices = np.arange(norm_max, dtype=np.float64)
    dat2 = hist * (indices + 1)

    total = dat2.sum()
    if total == 0:
        return None

    overall_mean = (indices * dat2).sum() / total

    weight_cumsum = np.cumsum(dat2)
    mean_cumsum = np.cumsum(indices * dat2)

    w1 = weight_cumsum / total
    m1 = np.zeros_like(w1)
    nz1 = weight_cumsum > 0
    m1[nz1] = mean_cumsum[nz1] / weight_cumsum[nz1]

    w2 = 1.0 - w1
    m2 = np.zeros_like(w2)
    nz2 = w2 > 0
    m2[nz2] = (overall_mean - w1[nz2] * m1[nz2]) / w2[nz2]

    bcv = w1 * w2 * (m1 - m2) ** 2
    t_best = int(np.argmax(bcv))

    threshold_size = min_val + (t_best / (norm_max - 1)) * (max_val - min_val)
    threshold_size = int(np.round(threshold_size))

    return threshold_size


# -------------------------------------
#  One image denoising
# -------------------------------------
def _denoise_one_image(
    img_gray,
    norm_max=p_norm_max,
    num_iterations=p_num_iterations,
    connectivity=8,
    size_scale=p_size_scale,
    min_size=p_min_size,
):
    """
    connected-component-based noise removal.
    回傳 denoised 的二值圖 (uint8, 0/255)
    """

    # Step 1: Otsu threshold
    _, bin_img = cv2.threshold(
        img_gray, 0, 255, cv2.THRESH_BINARY + cv2.THRESH_OTSU
    )

    # make sure text is black
    bin_img = _ensure_text_is_black(bin_img)

    for _ in range(num_iterations):
        fg = (bin_img == 0).astype(np.uint8)  # foreground = text = black

        num_labels, labels = cv2.connectedComponents(fg, connectivity=connectivity)
        if num_labels <= 1:
            break

        flat_labels = labels.ravel()
        sizes = np.bincount(flat_labels)
        component_sizes = sizes[1:]  # skip background

        threshold_size = _compute_otsu_threshold_on_component_sizes(
            component_sizes, norm_max=norm_max
        )

        if threshold_size is None or threshold_size <= 0:
            break

        # ⭐ 這裡開始是「溫和化」的關鍵調整 ⭐
        # 1) 用 size_scale 把 Otsu 的門檻縮小一些（少砍一點）
        threshold_size = int(threshold_size * size_scale)

        # 2) 至少要大於 min_size 才值得用來砍
        if threshold_size < min_size:
            threshold_size = min_size

        # 只刪掉「比 threshold 小的 component」
        small_labels = np.where(component_sizes < threshold_size)[0] + 1
        if small_labels.size == 0:
            break

        remove_mask = np.isin(labels, small_labels)
        bin_img[remove_mask] = 255

    return bin_img


# -------------------------------------
#  Folder-level processing (UPDATED)
# -------------------------------------
def denoise_document_folder(
    input_folder,
    output_root="Denoise_Doc",
    norm_max=5000,
    num_iterations=6,
    connectivity=8,
    verbose=True,
):
    """
    Noise Removal Technique for Document Images (full folder).

    NEW (updated to match median/bilateral outputs):
      ✔ Saves into Denoise_Doc/<dataset_name>/images/
      ✔ Copies ./datasets/<dataset_name>/gt.txt
      ✔ Returns (images folder, gt path)
    """

    input_folder = Path(input_folder)
    if not input_folder.is_dir():
        raise ValueError(f"Input folder does not exist: {input_folder}")

    # dataset name extracted: .../<dataset>/images
    name = input_folder.parts[-2]

    # create output structure
    output_root = Path(output_root)
    output_dataset_dir = output_root / name
    images_out_dir = output_dataset_dir / "images"
    images_out_dir.mkdir(parents=True, exist_ok=True)

    # copy gt.txt
    dataset_dir = input_folder.parent  # .../<dataset>
    gt_src = dataset_dir / "gt.txt"
    gt_dst = output_dataset_dir / "gt.txt"

    if gt_src.exists():
        shutil.copy2(gt_src, gt_dst)
    else:
        print(f"⚠️ gt.txt not found at: {gt_src}")

    # find images
    exts = (".png", ".jpg", ".jpeg", ".tif", ".tiff", ".bmp")
    image_paths = [
        p for p in sorted(input_folder.glob("*")) if p.suffix.lower() in exts
    ]

    if verbose:
        print(f"Found {len(image_paths)} images in {input_folder}")
        print(f"Output images → {images_out_dir}")

    iterator = tqdm(image_paths, desc="Document denoising", unit="img") if verbose else image_paths

    for img_path in iterator:
        img = cv2.imread(str(img_path), cv2.IMREAD_GRAYSCALE)
        if img is None:
            continue

        denoised_bin = _denoise_one_image(
            img,
            norm_max=norm_max,
            num_iterations=num_iterations,
            connectivity=connectivity,
        )

        out_path = images_out_dir / img_path.name
        cv2.imwrite(str(out_path), denoised_bin)

    if verbose:
        print("Document-image denoising completed.")

    return images_out_dir, gt_dst
```

```python
import os
import shutil
from pathlib import Path

import cv2
import pandas as pd

def build_ned_filtered_denoised_dataset_and_lmdb(
    dataset_name,
    detailed_csv_path,
    orig_dataset_root,
    out_root="Denoised_NED_Filtered",
    lmdb_root="lmdb",
    ned_threshold=1.0,
    denoise_params=None,
):
    """
    For a given dataset (e.g. 'MJ600' or 'D_prime'), do ALL of the following:

      1) Read detailed CSV with columns: 'idx' and 'norm_edit_distance'
      2) Create a new dataset at: out_root / dataset_name / images
            - NED == ned_threshold  → copy original image
            - NED <  ned_threshold  → denoise with `_denoise_one_image`
         Also copy gt.txt into out_root / dataset_name / gt.txt
      3) Call `create_lmdb` to create an LMDB at:
            lmdb_root / f"{dataset_name}_denoised"

    Returns:
      (out_dataset_dir, out_images_dir, lmdb_output_dir)
    """
#+++++++============================Edit Here=======================
    if denoise_params is None:
        denoise_params = dict(
            # norm_max=2000,
            # num_iterations=3,
            # connectivity=8,
            # size_scale=0.35,
            # min_size=4,

            norm_max=p_norm_max,
            num_iterations=p_num_iterations,
            connectivity=8,
            size_scale=p_size_scale,
            min_size=p_min_size,
        )

    dataset_name = str(dataset_name)
    detailed_csv_path = Path(detailed_csv_path)
    orig_dataset_root = Path(orig_dataset_root)

    orig_images_dir = orig_dataset_root / "images"
    orig_gt_path    = orig_dataset_root / "gt.txt"

    if not detailed_csv_path.is_file():
        raise FileNotFoundError(f"Detailed CSV not found: {detailed_csv_path}")
    if not orig_images_dir.is_dir():
        raise FileNotFoundError(f"Images folder not found: {orig_images_dir}")
    if not orig_gt_path.is_file():
        raise FileNotFoundError(f"gt.txt not found: {orig_gt_path}")

    # -----------------------------
    # Prepare output folders
    # -----------------------------
    out_root        = Path(out_root)
    out_dataset_dir = out_root / dataset_name
    out_images_dir  = out_dataset_dir / "images"

    out_images_dir.mkdir(parents=True, exist_ok=True)

    # Copy gt.txt
    gt_dst_path = out_dataset_dir / "gt.txt"
    shutil.copy2(orig_gt_path, gt_dst_path)
    print(f"[{dataset_name}] Copied gt.txt → {gt_dst_path}")

    # -----------------------------
    # Load detailed results
    # -----------------------------
    df_det = pd.read_csv(detailed_csv_path)
    if "idx" not in df_det.columns:
        raise KeyError(f"[{dataset_name}] Column 'idx' not found in CSV: {detailed_csv_path}")
    if "norm_edit_distance" not in df_det.columns:
        raise KeyError(f"[{dataset_name}] Column 'norm_edit_distance' not found in CSV.")

    df_det["idx"] = df_det["idx"].astype(int)

    # -----------------------------
    # Helper: find image by idx
    # -----------------------------
    def find_image_for_idx(idx: int) -> Path | None:
        """
        Find image whose filename starts with '<idx>_' inside orig_images_dir.
        Matches your dataset naming like '12_groundtruth.jpg'.
        """
        prefix = f"{idx}_"
        for fname in os.listdir(orig_images_dir):
            if fname.startswith(prefix):
                return orig_images_dir / fname
        return None

    # -----------------------------
    # Main loop: copy or denoise
    # -----------------------------
    n_copy = 0
    n_denoise = 0
    missing = []

    print(f"\n[{dataset_name}] Building NED-filtered dataset...")
    print(f"  CSV:      {detailed_csv_path}")
    print(f"  Orig img: {orig_images_dir}")
    print(f"  Out dir:  {out_images_dir}")
    print(f"  NED rule: == {ned_threshold} → copy, < {ned_threshold} → denoise\n")

    for _, row in df_det.iterrows():
        idx = int(row["idx"])
        ned = float(row["norm_edit_distance"])

        src_path = find_image_for_idx(idx)
        if src_path is None:
            missing.append(idx)
            continue

        dst_path = out_images_dir / src_path.name

        # NED == threshold → copy original
        if abs(ned - ned_threshold) < 1e-6:
            shutil.copy2(src_path, dst_path)
            n_copy += 1

        # NED < threshold → denoise
        elif ned < ned_threshold:
            img_gray = cv2.imread(str(src_path), cv2.IMREAD_GRAYSCALE)
            if img_gray is None:
                missing.append(idx)
                continue

            denoised_bin = _denoise_one_image(
                img_gray,
                **denoise_params,
            )
            cv2.imwrite(str(dst_path), denoised_bin)
            n_denoise += 1

        # NED > threshold (rare; just copy)
        else:
            shutil.copy2(src_path, dst_path)
            n_copy += 1

    print(f"\n=== [{dataset_name}] NED-filtered denoising summary ===")
    print(f"Output folder: {out_images_dir}")
    print(f"Copied original images (NED == {ned_threshold}): {n_copy}")
    print(f"Denoised images (NED <  {ned_threshold}): {n_denoise}")
    if missing:
        print(f"⚠️ Missing {len(missing)} images for indices: {missing[:20]}"
              + (" ..." if len(missing) > 20 else ""))

    # -----------------------------
    # Create LMDB using your helper
    # -----------------------------
    lmdb_output_dir = Path(lmdb_root) / f"{dataset_name}_filtered_denoised"

    create_lmdb(
        image_directory_original=str(out_dataset_dir),
        gt_file_original=str(gt_dst_path),
        lmdb_output_dir=str(lmdb_output_dir),
    )

    print(f"\n[{dataset_name}] LMDB created at: {lmdb_output_dir}\n")

    return out_dataset_dir, out_images_dir, lmdb_output_dir
```

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
import cv2
import random
import numpy as np

# =======================================================
# Adjustable parameters
# =======================================================
N_examples = 5     # number of random samples to display
M_improved = 5     # number of improved cases to display

dataset = "MJ600"  # or "D_prime"

# =======================================================
# Paths
# =======================================================
result_root   = Path("./result/None-VGG-None-CTC.pth")
data_root     = Path("datasets") / dataset / "images"
denoised_root = Path("Denoise_Doc") / dataset / "images"
filtered_root = Path("Denoised_NED_Filtered") / dataset / "images"

csv_orig     = result_root / f"{dataset}_detailed_results.csv"
csv_denoised = result_root / f"{dataset}_denoised_detailed_results.csv"
csv_filtered = result_root / f"{dataset}_filtered_denoised_detailed_results.csv"

# =======================================================
# Load CSVs & rename to avoid column clashes
# =======================================================
df_orig = pd.read_csv(csv_orig).rename(
    columns={
        "norm_edit_distance": "ned_orig",
        "prediction": "pred_orig",
        "correct": "correct_orig",
    }
)

df_denoised = pd.read_csv(csv_denoised).rename(
    columns={
        "norm_edit_distance": "ned_denoised",
        "prediction": "pred_denoised",
        "correct": "correct_denoised",
    }
)

df_filtered = pd.read_csv(csv_filtered).rename(
    columns={
        "norm_edit_distance": "ned_filtered",
        "prediction": "pred_filtered",
        "correct": "correct_filtered",
    }
)

# Merge on idx
df = df_orig.merge(
        df_denoised[["idx", "ned_denoised", "pred_denoised", "correct_denoised"]],
        on="idx", how="left"
    ).merge(
        df_filtered[["idx", "ned_filtered", "pred_filtered", "correct_filtered"]],
        on="idx", how="left"
    )

# =======================================================
# Helper to load an image
# =======================================================
def load_image(idx, root: Path):
    """Load image whose filename starts with '<idx>_' from given root."""
    if not root.exists():
        return None
    prefix = f"{idx}_"
    for f in os.listdir(root):
        if f.startswith(prefix):
            return cv2.imread(str(root / f), cv2.IMREAD_GRAYSCALE)
    return None

# =======================================================
# 1. Show N random examples (with GT only in suptitle)
# =======================================================
sample_indices = random.sample(list(df["idx"]), N_examples)

for idx in sample_indices:
    row = df[df["idx"] == idx].iloc[0]

    img_o = load_image(idx, data_root)
    img_d = load_image(idx, denoised_root)
    img_f = load_image(idx, filtered_root)

    plt.figure(figsize=(12, 4))

    # Original
    plt.subplot(1, 3, 1)
    if img_o is not None:
        plt.imshow(img_o, cmap='gray')
    plt.title(f"Original\nPred: {row['pred_orig']}\nNED={row['ned_orig']:.3f}")
    plt.axis("off")

    # Denoised
    plt.subplot(1, 3, 2)
    if img_d is not None:
        plt.imshow(img_d, cmap='gray')
    plt.title(f"Denoised\nPred: {row['pred_denoised']}\nNED={row['ned_denoised']:.3f}")
    plt.axis("off")

    # Filtered
    plt.subplot(1, 3, 3)
    if img_f is not None:
        plt.imshow(img_f, cmap='gray')
    plt.title(f"Filtered\nPred: {row['pred_filtered']}\nNED={row['ned_filtered']:.3f}")
    plt.axis("off")

    # Unified GT in suptitle
    plt.suptitle(
        f"Random Example idx={idx}\n"
        f"GT: {row['ground_truth']}"
    )
    plt.tight_layout(rect=[0, 0, 1, 0.90])
    plt.show()

# =======================================================
# 2. Show M improved cases (Original → Filtered)
# =======================================================
df["improved"] = df["ned_filtered"] > df["ned_orig"] + 1e-6
improved_cases = df[df["improved"]].sort_values(by="ned_filtered", ascending=False).head(M_improved)

print(f"\nFound {len(improved_cases)} improved cases. Showing up to {M_improved}.\n")

for _, row in improved_cases.iterrows():
    idx = row["idx"]

    img_o = load_image(idx, data_root)
    img_f = load_image(idx, filtered_root)

    plt.figure(figsize=(10, 4))

    # Original
    plt.subplot(1, 2, 1)
    if img_o is not None:
        plt.imshow(img_o, cmap='gray')
    plt.title(f"Original\nPred: {row['pred_orig']}\nNED={row['ned_orig']:.3f}")
    plt.axis("off")

    # Filtered
    plt.subplot(1, 2, 2)
    if img_f is not None:
        plt.imshow(img_f, cmap='gray')
    plt.title(f"Filtered\nPred: {row['pred_filtered']}\nNED={row['ned_filtered']:.3f}")
    plt.axis("off")

    # Unified GT in suptitle
    plt.suptitle(
        f"Improved Case idx={idx}\n"
        f"GT: {row['ground_truth']}"
    )
    plt.tight_layout(rect=[0, 0, 1, 0.90])
    plt.show()
```

```python
import os
import pandas as pd
import matplotlib.pyplot as plt
from pathlib import Path
import cv2

def show_idx_comparison(idx, dataset="MJ600"):
    """
    Show comparison of one sample (by idx) across:
      - Original dataset
      - Denoised dataset
      - Filtered-denoised dataset

    Assumes detailed CSVs have columns:
      idx, ground_truth, prediction, correct, norm_edit_distance

    And images live in:
      datasets/<dataset>/images
      Denoise_Doc/<dataset>/images
      Denoised_NED_Filtered/<dataset>/images
    """
    # ---------------------------------------------------
    # Paths
    # ---------------------------------------------------
    result_root   = Path("./result/None-VGG-None-CTC.pth")
    data_root     = Path("datasets") / dataset / "images"
    denoised_root = Path("Denoise_Doc") / dataset / "images"
    filtered_root = Path("Denoised_NED_Filtered") / dataset / "images"

    csv_orig     = result_root / f"{dataset}_detailed_results.csv"
    csv_denoised = result_root / f"{dataset}_denoised_detailed_results.csv"
    csv_filtered = result_root / f"{dataset}_filtered_denoised_detailed_results.csv"

    # ---------------------------------------------------
    # Load CSVs & rename to avoid clashes
    # ---------------------------------------------------
    df_orig = pd.read_csv(csv_orig).rename(
        columns={
            "norm_edit_distance": "ned_orig",
            "prediction": "pred_orig",
            "correct": "correct_orig",
        }
    )

    df_denoised = pd.read_csv(csv_denoised).rename(
        columns={
            "norm_edit_distance": "ned_denoised",
            "prediction": "pred_denoised",
            "correct": "correct_denoised",
        }
    )

    df_filtered = pd.read_csv(csv_filtered).rename(
        columns={
            "norm_edit_distance": "ned_filtered",
            "prediction": "pred_filtered",
            "correct": "correct_filtered",
        }
    )

    # Merge on idx
    df = df_orig.merge(
            df_denoised[["idx", "ned_denoised", "pred_denoised", "correct_denoised"]],
            on="idx", how="left"
        ).merge(
            df_filtered[["idx", "ned_filtered", "pred_filtered", "correct_filtered"]],
            on="idx", how="left"
        )

    # ---------------------------------------------------
    # Grab the row for this idx
    # ---------------------------------------------------
    if idx not in df["idx"].values:
        print(f"idx={idx} not found in dataset {dataset}.")
        return

    row = df[df["idx"] == idx].iloc[0]

    # ---------------------------------------------------
    # Helper: load image by prefix "<idx>_"
    # ---------------------------------------------------
    def load_image(idx, root: Path):
        if not root.exists():
            return None
        prefix = f"{idx}_"
        for f in os.listdir(root):
            if f.startswith(prefix):
                return cv2.imread(str(root / f), cv2.IMREAD_GRAYSCALE)
        return None

    img_o = load_image(idx, data_root)
    img_d = load_image(idx, denoised_root)
    img_f = load_image(idx, filtered_root)

    # ---------------------------------------------------
    # Plot
    # ---------------------------------------------------
    fig = plt.figure(figsize=(12, 4))

    # Original
    ax1 = fig.add_subplot(1, 3, 1)
    if img_o is not None:
        ax1.imshow(img_o, cmap='gray')
    ax1.set_title(
        f"Original\n"
        f"Pred: {row['pred_orig']}\n"
        f"NED={row['ned_orig']:.3f}"
    )
    ax1.axis("off")

    # Denoised
    ax2 = fig.add_subplot(1, 3, 2)
    if img_d is not None:
        ax2.imshow(img_d, cmap='gray')
    ax2.set_title(
        f"Denoised\n"
        f"Pred: {row['pred_denoised']}\n"
        f"NED={row['ned_denoised']:.3f}"
    )
    ax2.axis("off")

    # Filtered-denoised
    ax3 = fig.add_subplot(1, 3, 3)
    if img_f is not None:
        ax3.imshow(img_f, cmap='gray')
    ax3.set_title(
        f"Filtered\n"
        f"Pred: {row['pred_filtered']}\n"
        f"NED={row['ned_filtered']:.3f}"
    )
    ax3.axis("off")

    # GT only once in suptitle
    fig.suptitle(
        f"{dataset} — idx={idx}  |  GT: {row['ground_truth']}",
        y=0.98, fontsize=14
    )
    plt.subplots_adjust(top=0.7)
    plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# --- paths to your combined CSVs ---
csv_dprime = "result/None-VGG-None-CTC.pth/D_prime_combined.csv"
csv_mj600  = "result/None-VGG-None-CTC.pth/MJ600_combined.csv"

# --- load ---
df_dprime = pd.read_csv(csv_dprime)
df_mj600  = pd.read_csv(csv_mj600)

# --- label datasets ---
df_dprime["dataset"] = "D_prime"
df_mj600["dataset"]  = "MJ600"

# --- concatenate for plotting ---
df_all = pd.concat([df_dprime, df_mj600], ignore_index=True)

# --- columns to compare ---
metrics = ["sharp_score", "noise_score", "contrast_score", "low_light_score"]
titles  = ["Sharp", "Noise", "Contrast", "Low-light"]

# --- plot ---
sns.set(style="whitegrid")

for m, t in zip(metrics, titles):
    plt.figure(figsize=(8, 4))
    sns.kdeplot(
        data=df_all, x=m, hue="dataset", fill=True, common_norm=False,
        alpha=0.5, linewidth=1.2
    )
    plt.title(f"{t} Score Distribution: D_prime vs MJ600")
    plt.xlabel("Normalized score (0–1)")
    plt.ylabel("Density")
    plt.grid(True)
    plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# --- paths to your combined CSVs ---
csv_dprime = "result/None-VGG-None-CTC.pth/D_prime_combined.csv"
csv_mj600  = "result/None-VGG-None-CTC.pth/MJ600_combined.csv"

# --- load ---
df_dprime = pd.read_csv(csv_dprime)
df_mj600  = pd.read_csv(csv_mj600)

# --- label datasets ---
df_dprime["dataset"] = "D_prime"
df_mj600["dataset"]  = "MJ600"

# --- concatenate for plotting ---
df_all = pd.concat([df_dprime, df_mj600], ignore_index=True)

# --- columns to compare ---
performance_metrics = ["confidence", "norm_edit_distance"]
quality_metrics = ["sharp_score", "noise_score", "contrast_score", "low_light_score"]
quality_titles  = ["Sharpness", "Noise", "Contrast", "Low-light"]

# --- plot ---
sns.set(style="whitegrid")

for perf_metric in performance_metrics:
    for quality_metric, quality_title in zip(quality_metrics, quality_titles):
        plt.figure(figsize=(8, 6))
        sns.regplot(
            data=df_dprime,
            x=quality_metric,
            y=perf_metric,
            scatter_kws={'alpha':0.6, 's':15},
            label="D_prime",
            ci=95  # Confidence interval
        )
        sns.regplot(
            data=df_mj600,
            x=quality_metric,
            y=perf_metric,
            scatter_kws={'alpha':0.6, 's':15},
            label="MJ600",
            ci=95 # Confidence interval
        )
        plt.title(f"Correlation: {perf_metric.replace('_', ' ').title()} vs {quality_title}")
        plt.xlabel(f"{quality_title} Score (0-1)")
        plt.ylabel(perf_metric.replace('_', ' ').title())
        plt.legend()
        plt.grid(True)
        plt.show()
```
